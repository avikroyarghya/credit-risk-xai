# 02 — Exploratory Data Analysis

## Goal
Univariate, bivariate and target-level exploration: what separates defaulting clients from the rest?

```python
from pathlib import Path
import sys

# Locate the project root (works whether the notebook is run from notebooks/ or root)
root = Path.cwd()
while not (root / "src" / "data_loader.py").exists() and root != root.parent:
    root = root.parent
sys.path.insert(0, str(root))
print("project root:", root)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from src.data_loader import load_clean, TARGET

df = load_clean()
print(df.shape)
df[TARGET].value_counts(normalize=True).round(3)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4), constrained_layout=True)
sns.kdeplot(data=df, x="LIMIT_BAL", hue=TARGET, common_norm=False, fill=True, ax=axes[0])
axes[0].set_title("Credit limit by outcome")
rate = df.groupby("PAY_0")[TARGET].mean()
rate.plot(kind="bar", color="#E76F51", ax=axes[1], title="Default rate by Sep repayment status")
axes[1].set_xlabel("PAY_0 status code")
plt.show()
```

```python
num = ["LIMIT_BAL", "AGE", "PAY_0", "BILL_AMT1", "PAY_AMT1", TARGET]
df[num].corr()[TARGET].sort_values()
```

The full figure set (13 plots) is produced by `python -m src.eda` and saved to `reports/figures/`. Highlights: default rate rises from 17% for 'paid duly' to 69% at a 2-month delay; higher credit limits are associated with lower default rates.
