**Findings**
- For logistic regression, plain `class_weight='balanced'` or SMOTE *hurt* precision/recall balance at the default threshold (they trade precision for recall) — the 'none' pipeline wins on PR-AUC.
- For XGBoost, SMOTE adds only ~0.002 PR-AUC over no handling.
- Takeaway: moderate imbalance (22%) is often best handled by **threshold tuning** rather than resampling; always decide on a validation set.
